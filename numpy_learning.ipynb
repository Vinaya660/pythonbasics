```python
# 1. Create classification target
df["price_category"] = (df["price"] >= 30).astype(int)

# 2. X and y
X1 = df.drop(columns=["price", "price_category"])
y1 = df["price_category"]



# 4. Train-test split
from sklearn.model_selection import train_test_split

x1_train, x1_test, y1_train, y1_test = train_test_split(
    X1,
    y1,
    test_size=0.2,
    random_state=42,
    stratify=y1
)

```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
model=lr.fit(x1_train,y1_train)
```

```python
y_pred=model.predict(x1_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y1_test,y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
dr=DecisionTreeClassifier()
```

```python
model=dr.fit(x1_train,y1_train)
y_pred=model.predict(x1_test)
```

```python
accuracy_score(y1_test,y_pred)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y1_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y1_test, y_pred))
```
